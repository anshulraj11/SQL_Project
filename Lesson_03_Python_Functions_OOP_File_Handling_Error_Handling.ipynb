# 1. Functions & Argument Matching

A function can accept parameters using:
- **Positional arguments**
- **Keyword arguments**
- **Default arguments**

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

# Positional arguments
print(greet("Alice", "Good morning"))

# Keyword arguments
print(greet(name="Alice", greeting="Good morning"))

# Default argument
print(greet("Alice"))
```

# 2. Arbitrary Arguments

Use:
- `*args` for any number of positional arguments.
- `**kwargs` for any number of keyword arguments.

```python
def display_items(*args, **kwargs):
    print("Positional Arguments:")
    for i, item in enumerate(args, start=1):
        print(f"  Item {i}: {item}")

    print("\nKeyword Arguments:")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

display_items(
    "Apple", "Banana", "Cherry",
    name="Alice", age=30, city="New York"
)
```

# 3. `return` Statement

`return` sends a value from a function back to the place where the function was called.

```python
def calculate_sum(a, b):
    return a + b

result = calculate_sum(10, 20)
print(f"The sum is: {result}")
```

# 4. Generators, Lambda & Built-in Functions

### Generator
A generator uses `yield` to produce values one at a time.

### Lambda
A lambda is a small anonymous function.

### `map()` and `filter()`
- `map()` applies a function to every item.
- `filter()` keeps items that satisfy a condition.

```python
# Generator
def number_generator(n):
    for i in range(1, n + 1):
        yield i

print("Generated numbers:")
for num in number_generator(5):
    print(num)

# Lambda
square = lambda x: x ** 2
print("\nSquare of 4:", square(4))

# map() and filter()
numbers = [1, 2, 3, 4, 5]

squared_numbers = list(map(lambda x: x ** 2, numbers))
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Squared:", squared_numbers)
print("Even:", even_numbers)
```

# 5. Class, Attributes & Methods

A class is a blueprint for creating objects.
- **Attributes** store information.
- **Methods** define behavior.
- Objects are created from the class.

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Car: {self.year} {self.brand} {self.model}")

car1 = Car("Toyota", "Corolla", 2022)
car2 = Car("Honda", "Civic", 2023)

car1.display_info()
car2.display_info()
```

# 6. Inheritance

**Inheritance** allows a child class to reuse attributes and methods from a parent class.

Here, `Dog` inherits from `Animal` and adds its own `fetch()` method.

```python
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        print(f"{self.name} the {self.species} makes a sound.")


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "Dog")
        self.breed = breed

    def fetch(self):
        print(f"{self.name} is fetching the ball!")


dog1 = Dog("Buddy", "Golden Retriever")

print(f"Name: {dog1.name}")
print(f"Species: {dog1.species}")
print(f"Breed: {dog1.breed}")

dog1.make_sound()   # Inherited method
dog1.fetch()        # Child method
```

# 7. File Handling

Python can:
- Write to a file using `"w"`
- Append using `"a"`
- Read using `"r"`
- Read line by line

The `with open(...)` pattern automatically manages the file after the block finishes.

```python
file_name = "example.txt"

# Write (creates or replaces the file)
with open(file_name, "w") as file:
    file.write("This is the first line.\n")
    file.write("Writing to a file in Python.\n")

# Append (adds to existing content)
with open(file_name, "a") as file:
    file.write("Appending a new line.\n")
    file.write("Another appended line.\n")

# Read the complete file
with open(file_name, "r") as file:
    print("Complete file:")
    print(file.read())

# Read line by line
print("Line by line:")
with open(file_name, "r") as file:
    for line in file:
        print(line.strip())
```

### Optional: Manual File Closing

The original demonstration also shows manual closing with `close()`. The `with` approach above is generally simpler for beginners because the file is closed automatically.

```python
file = open("example.txt", "w")
file.write("Final content replacing everything.\n")
file.write("Closing the file explicitly.\n")
file.close()

print("File closed successfully.")
```

# 8. Error Handling

Error handling prevents a program from stopping unexpectedly.

The uploaded demonstration focuses on:
- `try` — code that may cause an error
- `except FileNotFoundError` — handles a missing file
- `with open()` — manages the file automatically

```python
file_name = input("Enter the file name: ")

try:
    with open(file_name, "r") as file:
        print("\nFile Contents:\n")
        print(file.read())

except FileNotFoundError:
    print(f"\nError: The file '{file_name}' was not found.")

print("\nOperation completed.")
```
